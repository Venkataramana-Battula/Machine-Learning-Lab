```python
import numpy as np

```

```python
def multiclass_metrics(conf_matrix):
    conf_matrix = np.array(conf_matrix)
    n_classes = conf_matrix.shape[0]

    total = np.sum(conf_matrix)

    accuracy = np.trace(conf_matrix) / total

    precision_list = []
    recall_list = []
    f1_list = []

    for i in range(n_classes):
        TP = conf_matrix[i, i]
        FP = np.sum(conf_matrix[:, i]) - TP
        FN = np.sum(conf_matrix[i, :]) - TP
        TN = total - (TP + FP + FN)

        precision = TP / (TP + FP) if (TP + FP) != 0 else 0
        recall = TP / (TP + FN) if (TP + FN) != 0 else 0

        f1 = (2 * precision * recall) / (precision + recall) if (precision + recall) != 0 else 0

        precision_list.append(precision)
        recall_list.append(recall)
        f1_list.append(f1)

        print(f"\nClass {i}:")
        print("TP:", TP, "FP:", FP, "FN:", FN, "TN:", TN)
        print("Precision:", precision)
        print("Recall:", recall)
        print("F1 Score:", f1)

    # Macro Average
    print("\nOverall Metrics:")
    print("Accuracy:", accuracy)
    print("Macro Precision:", np.mean(precision_list))
    print("Macro Recall:", np.mean(recall_list))
    print("Macro F1 Score:", np.mean(f1_list))


```

```python
# Example Confusion Matrix (3-class)
conf_matrix = [
    [50, 2, 3],
    [5, 45, 5],
    [2, 3, 40]
]

multiclass_metrics(conf_matrix)
```
