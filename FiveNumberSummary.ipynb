```python
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt

```

```python
# Given NumPy array (example data)
data = np.array([12, 15, 14, 10, 18, 22, 25, 19, 30, 28, 16, 14, 13, 17, 21])

```

```python
# Compute Five-Number Summary
minimum = np.min(data)
q1 = np.percentile(data, 25)
median = np.median(data)
q3 = np.percentile(data, 75)
maximum = np.max(data)

five_number_summary = {
    "Minimum": minimum,
    "Q1 (25th Percentile)": q1,
    "Median (Q2)": median,
    "Q3 (75th Percentile)": q3,
    "Maximum": maximum
}

print("Five-Number Summary:")
for key, value in five_number_summary.items():
    print(f"{key}: {value}")

```

```python
# Plot Box Plot
plt.figure()
plt.boxplot(data)
plt.title("Box Plot")
plt.ylabel("Values")
plt.show()
```
